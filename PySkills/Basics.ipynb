### Wednesday: Working with Strings


```python
name = "John" + " " + "Smith"

name
```

```python
fname = "John"
lname = 'Smith'

full_name = fname + " " + lname

full_name
```

```python
# Formatting Strings

# .format()

location = 'Lagos'

print('Hello {}, I trust you are now in {}!'.format(fname, location))
```

```python
# f() Method

print(f"Hello {fname}, I trust you are now in {location}!")
```

```python
# Python 2

print("Hello, %s %s" % (fname, lname))
```

```python
# String Index

print(len(fname))
```

```python
print(fname[0], fname[1], fname[2], fname[3])
```

```python
print(fname[-4])
```

```python
# String Slicing (Syntax: var_name[start:stop:step])

word = 'Hello'

word[0:5:2]
```

```python
sentence = "I love Nigeria and Canada"

len(sentence)
```

```python
for i in range(len(sentence)):
    print(sentence[i])
```
